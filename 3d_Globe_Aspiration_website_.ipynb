```python
# ================================================================
# CELL 4: BUILD THE PAGE AND OPEN IT
# ================================================================
import json
import base64
import urllib.request
import webbrowser
import pathlib

# 1. Get the world land outlines (downloaded once, then reused)
if not os.path.exists(land_file):
    print("Downloading land outlines...")
    land_url = "https://cdn.jsdelivr.net/npm/world-atlas@2.0.2/land-50m.json"
    urllib.request.urlretrieve(land_url, land_file)
with open(land_file, "r", encoding="utf-8") as f:
    land_text = f.read()

# 2. Read the CEPT logo, trim its white border, and turn it into text for the HTML
logo_src = ""
if os.path.exists(logo_file):
    try:
        from PIL import Image, ImageChops
        import io
        logo = Image.open(logo_file).convert("RGB")
        # find where the logo is by comparing it with a plain white image
        white = Image.new("RGB", logo.size, (255, 255, 255))
        table = []
        for v in range(256):
            if v > 20:
                table.append(255)
            else:
                table.append(0)
        diff = ImageChops.difference(logo, white).convert("L").point(table)
        box = diff.getbbox()
        if box:
            pad = 15
            left = max(box[0] - pad, 0)
            top = max(box[1] - pad, 0)
            right = min(box[2] + pad, logo.width)
            bottom = min(box[3] + pad, logo.height)
            logo = logo.crop((left, top, right, bottom))
        buffer = io.BytesIO()
        logo.save(buffer, format="PNG")
        logo_bytes = buffer.getvalue()
        logo_type = "image/png"
    except ImportError:
        # Pillow not installed: use the file as it is (white border stays)
        with open(logo_file, "rb") as f:
            logo_bytes = f.read()
        if logo_file.lower().endswith(".png"):
            logo_type = "image/png"
        else:
            logo_type = "image/jpeg"
    logo_src = "data:" + logo_type + ";base64," + base64.b64encode(logo_bytes).decode("ascii")
    print("Logo added:", logo_file)
else:
    print("Logo not found at", logo_file, "- the page will be built without it")

# 3. Put all the data together
page_data = {
    "title1": title_line_1,
    "title2": title_line_2,
    "group": group_name,
    "home": home_city,
    "colours": colours,
    "descriptions": descriptions,
    "locations": locations,
    "indicator_cities": indicator_cities,
    "benchmarks": benchmarks,
    "acr_context": acr_context,
    "data_notes": data_notes,
    "recommendations": recommendations,
}
data_text = json.dumps(page_data, ensure_ascii=False)

# 4. Fill the template and save the page
page = html_template
page = page.replace("__DATA_JSON__", data_text)
page = page.replace("__LAND_JSON__", land_text)
page = page.replace("__LOGO_SRC__", logo_src)

with open(html_file, "w", encoding="utf-8") as f:
    f.write(page)

print("Saved:", html_file)
webbrowser.open(pathlib.Path(html_file).resolve().as_uri())
```

```python
# ================================================================
# CELL 5 (OPTIONAL): SAVE A ROTATING GIF FOR POWERPOINT
# First time only, run these two lines in a new cell, then restart the kernel:
#   %pip install playwright pillow
#   !playwright install chromium
# ================================================================
import subprocess
import sys

gif_file = os.path.join(folder, "global_aspiration_globe.gif")
gif_frames = 72          # more frames = smoother but bigger file
gif_step = 5             # degrees turned per frame (72 x 5 = one full turn)
gif_start_lon = 72.6     # start the turn at Ahmedabad
gif_lat = 15             # tilt of the globe
gif_zoom = 1.0           # 1 = normal size
gif_ms = 70              # time per frame in milliseconds
gif_width = 1100
gif_height = 800

# This small program runs in its own Python process.
# (Playwright cannot start inside Jupyter's event loop on Windows.)
worker = r'''
import sys, io, pathlib
from PIL import Image
from playwright.sync_api import sync_playwright

html, out = sys.argv[1], sys.argv[2]
frames, step, lon0, lat, zoom, ms, w, h = [float(x) for x in sys.argv[3:11]]
url = pathlib.Path(html).resolve().as_uri()

with sync_playwright() as p:
    browser = p.chromium.launch()
    page = browser.new_page(viewport={"width": int(w) + 460, "height": int(h) + 101}, device_scale_factor=1)
    page.goto(url)
    page.wait_for_timeout(2500)
    page.add_style_tag(content=".controls,.hint,#tip{display:none!important}")
    images = []
    for i in range(int(frames)):
        page.evaluate("([a, b, c]) => window.__globe.setView(a, b, c)", [lon0 - i * step, lat, zoom])
        page.wait_for_timeout(40)
        png = page.locator("#stage").screenshot()
        images.append(Image.open(io.BytesIO(png)).convert("RGB").quantize(colors=256, method=Image.Quantize.FASTOCTREE))
    browser.close()

images[0].save(out, save_all=True, append_images=images[1:], duration=int(ms), loop=0)
print("frames:", len(images))
'''

result = subprocess.run(
    [sys.executable, "-c", worker, html_file, gif_file,
     str(gif_frames), str(gif_step), str(gif_start_lon), str(gif_lat),
     str(gif_zoom), str(gif_ms), str(gif_width), str(gif_height)],
    capture_output=True, text=True)

print(result.stdout)
if result.returncode != 0:
    print("Something went wrong:")
    print(result.stderr)
else:
    print("GIF saved:", gif_file)
```
